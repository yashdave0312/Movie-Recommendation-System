```python
df = df[["title","overview","genres","tagline","vote_average","popularity"]]
df
```

```python
df = df.dropna(subset = ["title"])
```

```python
df["overview"] = df["overview"].fillna(" ")
```

```python
import ast
df["genres"] = df["genres"].apply(lambda x :" ".join([i["name"] for i in ast.literal_eval(x)]))
```

```python
df["tagline"] = df["tagline"].fillna(" ")
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df["tags"] = df["overview"] + df["genres"] + df["tagline"]
```

```python
df.head()
```

```python
# Text cleaning using NLP
# 1) converting to lower
df["tags"] = df["tags"].str.lower()
```

```python
# 2)Removing Punctuations
import string

def remove_punctuation(text):
    return text.translate(str.maketrans("" , "" , string.punctuation))

df["tags"] = df["tags"].apply(remove_punctuation)
df.head()
```

```python
# 3) Removing Numbers

def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['tags'] = df['tags'].apply(remove_numbers)        
```

```python
# Removing stopwords 

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
```

```python
df = df.reset_index(drop=True)
```

```python
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

for i in range(len(df)):
    words = word_tokenize(df.loc[i, "tags"])
    words = [lemmatizer.lemmatize(word) for word in words if word not in stop_words]
    df.loc[i, "tags"] = " ".join(words)

# def clean_tags(text):
#     words = word_tokenize(text)

#     words = [
#         lemmatizer.lemmatize(word)
#         for word in words
#         if word.lower() not in stop_words
#     ]

#     return " ".join(words)

# df["tags"] = df["tags"].apply(clean_tags)
```

```python
# creating the indices of the title 
df = df.reset_index(drop=True)
indices = pd.Series(df.drop_duplicates("title").index,index=df.drop_duplicates("title")["title"])
indices.head()
```

```python
# Converting the text into vectors 

from sklearn.feature_extraction.text import TfidfVectorizer 

tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words="english")

tfidf_matrix = tfidf.fit_transform(df["tags"])
```

```python
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommended(title, n=10):
    if title not in indices:
        return ["Movie not found"]

    idx = indices[title]
    sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix)
    sim_idx = sim_score.argsort()[0][::-1][1:n+1]
    
    return df["title"].iloc[sim_idx]

# Call the function correctly with a string
recommended("Toy Story")
```
